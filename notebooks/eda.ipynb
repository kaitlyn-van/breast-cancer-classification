```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# file path
sys.path.append(os.path.abspath(".."))

# sklearn dataset
from utils import load_data
```

```python
# load dataset (wisconsin breast cancer)
df = load_data()

print(df.head())
print(list(df.columns))

target = df["target"]
print(list(df["target"])) # malignant (0) or benign (1)

# shape of dataset
print(df.shape)

# distribution of target variable
print(target.value_counts())
```

```python
# bar chart visualization of target variable
categories = ['Malignant', 'Benign']
counts = np.bincount(target)

plt.figure(figsize=(8,6))
plt.bar(
    x=categories,
    height=counts,
    color=['tab:red', 'tab:blue']
)
plt.title("Tumour Distribution")
plt.xlabel("Target")
plt.ylabel("Counts")
plt.show()
```

```python
# seaborn heat map
plt.figure(figsize=(14,10))
plt.title("Feature Correlation Heatmap")
sns.heatmap(
    data=df.corr(),
    cmap='coolwarm',
    center=0
)
plt.show()
```

```python
# seaborn pairplot
pairplot = sns.pairplot(
            data=df,
            vars=['mean radius', 'mean texture', 'mean perimeter',
                'mean area', 'mean smoothness', 'mean concavity',
                'mean concave points', 'mean fractal dimension' 
            ],
            hue='target',
)
```
