```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# sklearn dataset
from sklearn.datasets import load_breast_cancer
```

```python
# load dataset (wisconsin breast cancer)
data = load_breast_cancer(as_frame=True)
df = data.frame

print(df.head())
print(list(df.columns))

target = df["target"]
print(data.target_names) # malignant (1) or benign (0)

# shape of dataset
print(df.shape)

# distribution of target variable
print(target.value_counts())
```

```python
# bar chart visualization of target variable
categories = ['Benign', 'Malignant']
counts = np.bincount(target)

plt.figure(figsize=(8,6))
plt.bar(
    x=categories,
    height=counts,
    color=['tab:blue', 'tab:red']
)
plt.title("Tumour Distribution")
plt.xlabel("Target")
plt.ylabel("Counts")
plt.show()
```

```python
# seaborn heat map
plt.figure(figsize=(14,10))
plt.title("Feature Correlation Heatmap")
sns.heatmap(
    data=df.corr(),
    cmap='coolwarm',
    center=0
)
plt.show()
```

```python
# seaborn pairplot
pairplot = sns.pairplot(
            data=df,
            vars=['mean radius', 'mean texture', 'mean perimeter',
                'mean area', 'mean smoothness', 'mean concavity',
                'mean concave points', 'mean fractal dimension' 
            ],
            hue='target',
)
```
